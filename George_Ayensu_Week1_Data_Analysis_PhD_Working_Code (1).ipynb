```python
import pyomo.environ as pyo
# 1. Creat the model
model = pyo.ConcreteModel()
# 2. Define the variables (How much each pump runs)
model.pump_a = pyo.Var(within=pyo.NonNegativeReals)
model.pump_b = pyo.Var(within=pyo.NonNegativeReals)
# 3. Define the Constraint(The limits)
model.limit_a = pyo.Constraint(expr=model.pump_a <= 40)
model.limit_b = pyo.Constraint(expr=model.pump_b <= 50)
# We must meet the demand of exactly 100.
model.demand = pyo.Constraint(expr=model.pump_a + model.pump_b == 90)
# 4. Define the Objective (Minimize cost)
model.cost = pyo.Objective(
    expr=10 * model.pump_a + 20 * model.pump_b,
    sense=pyo.minimize
)
# 5. Solve the model
solver = pyo.SolverFactory('glpk')
results = solver.solve(model)
# 6. Print the results
print(f"Status: {results.solver.status}")
print(f"Optimal Cost: ${model.cost.expr()}")
print(f"Run Pump A: {model.pump_a.value} MWh")
print(f"Run Pump B: {model.pump_b.value} MWh")
```

```python
import pyomo.environ as pyo
# 1. Setup
model = pyo.ConcreteModel()
# 2. Define Time (a set of numbersfrom 1 to 24)
model.T = pyo.Set(initialize=range(1, 25))
# 3. Define the Variables for EVERY hour (t)
model.pump = pyo.Var(model.T, within=pyo.NonNegativeReals)
model.tank = pyo.Var(model.T, within=pyo.NonNegativeReals)
# 4. Define the Parameters (Prices and Demand for every hour)
# Let's pretend it's cheap at night (hour 1-6) and expensive during the day.
def price_rule(model, t):
  if t <=6 or t >= 22: return 5 # off-peak
  else: return 20               # peak
  model.price = pyo.Param(model.T, initialize=price_rule)
  # The city needs 5 units of water every hour
  model.demand = pyo.Param(model.T, initialize=5)
  # 5. Define Constraints
  # Rule A: Max pump capacity is 10 per hour
  model.max_pump = pyo.Constraint(model.T, rule=lambda m, t: m.pump[t] <= 10)
  # Rule B: Max tank capacity is 50
  model.max_tank = pyo.Constraint(model.T, rule=lambda m, t: m.tank[t] <= 50)
  # Rule C: The Hydraulic Balance (The most important rule)
  # Tank(t) =Tank(t-1) + Pump(t) -Demand(t)
  def tank_balance_rule(model, t):
    if t == 1:
        return model.tank[t] == 0 + model.pump[t] - model.demand[t]
    else:
        return model.tank[t] == model.tank[t-1] + model.pump[t] - model.demand[t]
model.balance = pyo.Constraint(model.T, rule=tank_balance_rule)
# 6. Define the Objective (Minimize total cost across all 24 hours)
model.cost = pyo.Objective(
          expr=sum(model.price[t] * model.pump[t] for t in model.T),
          sense=pyo.minimize
      )

# 7. Solve
solver = pyo.SolverFactory('glpk')
results = solver.solve(model)
# 8. Print Results
print(f"Status: {results.solver.status}")
print(f"Total Daily Cost: ${model.cost.expr()}")
print(".........................")
print("Hour | Price | Pump | Tank Level")
for t in model.T:
  print(f"{t:4} | ${model.price[t]:4} | {model.pump[t].value:4.1f} | {model.tank[t].value:4.1f}")
```

```python
# Author: George Ayensu, PE
# Project:PhD Working Code - GWL Operation
# Date: 13th November 2026

!pip install pyomo highspy -q
import pyomo.environ as pyo

model = pyo.ConcreteModel()
model.T = pyo.Set(initialize=range(1, 25))

# --- REAL-WORLD DATA ---
max_old = 3000
old_intensity = 0.533
max_new = 6750
new_intensity = 0.398

# CLEAR WELL CAPACITIES (Corrected Topology)
# Old Plant supplies 3 wells (3 x 4,546 m3)
max_well_old = 13638
# New Plant supplies 2 wells (2 x 4,546 m3)
max_well_new = 9092

# Variables
model.pump_old = pyo.Var(model.T, within=pyo.NonNegativeReals)
model.pump_new = pyo.Var(model.T, within=pyo.NonNegativeReals)
model.well_old = pyo.Var(model.T, within=pyo.NonNegativeReals)
model.well_new = pyo.Var(model.T, within=pyo.NonNegativeReals)

# Binary variables (Fixed speed pumps ON/OFF)
model.old_on = pyo.Var(model.T, within=pyo.Binary)
model.new_on = pyo.Var(model.T, within=pyo.Binary)

# Tariff & Demand
def price_rule(model, t):
    if t <= 6 or t >= 22: return 0.10 # Off-peak
    else: return 0.25 # Peak
model.price = pyo.Param(model.T, initialize=price_rule)

def demand_rule(model, t):
    if t >= 22 or t <= 4: return 6000 # Night drop
    else: return 8000 # Day peak
model.demand = pyo.Param(model.T, initialize=demand_rule)

# Constraints
model.limit_old = pyo.Constraint(model.T, rule=lambda m, t: m.pump_old[t] <= max_old * m.old_on[t])
model.limit_new = pyo.Constraint(model.T, rule=lambda m, t: m.pump_new[t] <= max_new * m.new_on[t])

# Fixed Speed (100% capacity or 0)
model.fixed_old = pyo.Constraint(model.T, rule=lambda m, t: m.pump_old[t] == max_old * m.old_on[t])
model.fixed_new = pyo.Constraint(model.T, rule=lambda m, t: m.pump_new[t] == max_new * m.new_on[t])

# Clear Well Max Capacity (Corrected)
model.well_cap_old = pyo.Constraint(model.T, rule=lambda m, t: m.well_old[t] <= max_well_old)
model.well_cap_new = pyo.Constraint(model.T, rule=lambda m, t: m.well_new[t] <= max_well_new)

# The Hydraulic Balance
# Note: We assume the city demand is split 30% from Old Plant wells, 70% from New Plant wells
def well_balance_old(m, t):
    if t == 1:
        return m.well_old[t] == 3000 + m.pump_old[t] - (m.demand[t] * 0.3)
    else:
        return m.well_old[t] == m.well_old[t-1] + m.pump_old[t] - (m.demand[t] * 0.3)
model.balance_old = pyo.Constraint(model.T, rule=well_balance_old)

def well_balance_new(m, t):
    if t == 1:
        return m.well_new[t] == 2000 + m.pump_new[t] - (m.demand[t] * 0.7)
    else:
        return m.well_new[t] == m.well_new[t-1] + m.pump_new[t] - (m.demand[t] * 0.7)
model.balance_new = pyo.Constraint(model.T, rule=well_balance_new)

# Objective
model.cost = pyo.Objective(
expr=sum(model.price[t] * ((model.pump_old[t] * old_intensity) + (model.pump_new[t] * new_intensity)) for t in model.T),
    sense=pyo.minimize
)

# Solve
solver = pyo.SolverFactory('appsi_highs')
solver.solve(model)

print(f"Total Daily Cost (Corrected Topology): ${model.cost.expr():.2f}")
print("-------------------------------------------------------------------------")
print("Hour | Price | Pump Old | Well Old (3) | Pump New | Well New (2)")
for t in model.T:
    print(f"{t:4} | ${model.price[t]:4.2f} | {model.pump_old[t].value:8.1f} | {model.well_old[t].value:12.1f} | {model.pump_new[t].value:8.1f} | {model.well_new[t].value:12.1f}")
```

```python
# ==========================================
# Author: George Ayensu, PE
# Project: Techno-Economic Assessment (TEA)
# Case Study: GWL Weija Head Works Storage Investment
# Date: November 15, 2026
# ==========================================

# 1. The Financial Inputs
# Baseline Cost: $14,817.00 | What-If Cost: $14,414.02
daily_savings = 14817.00 - 14414.02
annual_savings = daily_savings * 365

# CAPEX (Capital Expenditure) Estimate for 50,000 m3 Storage
capex_tank = 2000000  # Estimated $2,000,000 USD
discount_rate = 0.08  # 8% discount rate (Standard for infrastructure)
years = 20            # 20-year project lifespan

print("--- TECHNO-ECONOMIC ASSESSMENT (TEA) ---")
print(f"Date of Analysis: November 15, 2026")
print(f"Daily Energy Savings: ${daily_savings:,.2f}")
print(f"Annual Energy Savings: ${annual_savings:,.2f}")
print(f"Estimated CAPEX for Storage: ${capex_tank:,.2f}")

# 2. Simple Payback Period
payback_years = capex_tank / annual_savings
payback_months = (payback_years - int(payback_years)) * 12
print(f"\nSimple Payback Period: {int(payback_years)} years and {int(payback_months)} months")

# 3. Net Present Value (NPV) Calculation
npv = -capex_tank  # Initial investment
for year in range(1, years + 1):
    discounted_savings = annual_savings / ((1 + discount_rate) ** year)
    npv += discounted_savings

print(f"Net Present Value (NPV) over 20 years: ${npv:,.2f}")

if npv > 0:
    print("CONCLUSION: The storage investment is FINANCIALLY VIABLE.")
else:
    print("CONCLUSION: The storage investment is NOT financially viable at this CAPEX.")
```

```python
# ==========================================
# Author: George Ayensu, PE
# Project: Tier 3 - VFD Retrofit + Storage Scenario
# Case Study: GWL Weija Head Works
# Date: November 15, 2026
# ==========================================

# --- 1. THE AFFINITY LAWS (VFD MATH) ---
# Affinity Law: Power is proportional to the cube of the speed.
# Old Plant: 20% speed reduction (due to gland packing constraints)
old_vfd_savings_factor = 1 - (0.80 ** 3) # 48.8% energy savings
# New Plant: 40% speed reduction (modern seals allow deeper throttling)
new_vfd_savings_factor = 1 - (0.60 ** 3) # 78.4% energy savings

print("--- TIER 3: VFD RETROFIT ANALYSIS ---")
print(f"Old Plant VFD Energy Reduction: {old_vfd_savings_factor * 100:.1f}%")
print(f"New Plant VFD Energy Reduction: {new_vfd_savings_factor * 100:.1f}%")

# --- 2. THE NEW DAILY ENERGY COST (With VFDs) ---
# Let's assume the pumps run at an average of 15 hours a day at full load,
# and the rest of the time they are throttled by the VFD.
# (For a precise PhD model, this would be inside the Pyomo time-series,
# but for a quick TEA estimate, we apply the factor to the baseline cost).

baseline_daily_cost = 14817.00

# Assuming 60% of the baseline cost comes from the Old Plant and 40% from the New Plant
old_plant_daily_cost = baseline_daily_cost * 0.60
new_plant_daily_cost = baseline_daily_cost * 0.40

# Calculate new costs with VFD savings applied
new_old_cost = old_plant_daily_cost * (1 - old_vfd_savings_factor)
new_new_cost = new_plant_daily_cost * (1 - new_vfd_savings_factor)

tier3_daily_cost = new_old_cost + new_new_cost
tier3_daily_savings = baseline_daily_cost - tier3_daily_cost
tier3_annual_savings = tier3_daily_savings * 365

print(f"\nBaseline Daily Cost: ${baseline_daily_cost:,.2f}")
print(f"Tier 3 Daily Cost (With VFDs): ${tier3_daily_cost:,.2f}")
print(f"Tier 3 Daily Savings: ${tier3_daily_savings:,.2f}")
print(f"Tier 3 Annual Savings: ${tier3_annual_savings:,.2f}")

# --- 3. THE NEW TEA (Financial Viability) ---
# CAPEX: $2M for Tank + $1.5M for VFDs (Medium Voltage VFDs are expensive)
capex_tier3 = 3500000
discount_rate = 0.08
years = 20

npv_tier3 = -capex_tier3 + sum((tier3_annual_savings / ((1 + discount_rate) ** year)) for year in range(1, years + 1))
payback_years = capex_tier3 / tier3_annual_savings

print(f"\nEstimated CAPEX (Tank + VFDs): ${capex_tier3:,.2f}")
print(f"Simple Payback Period: {payback_years:.1f} years")
print(f"Net Present Value (20 years): ${npv_tier3:,.2f}")

if npv_tier3 > 0:
    print("CONCLUSION: The combined Storage + VFD investment is FINANCIALLY VIABLE.")
else:
    print("CONCLUSION: The combined Storage + VFD investment is NOT financially viable.")
```

```python
# ==========================================
# Author: George Ayensu, PE
# Project: Tier 3 TEA (Realistic Engineering Adjustments)
# Case Study: GWL Weija Head Works
# Date: November 15, 2026
# ==========================================

# 1. THEORETICAL VFD SAVINGS (Affinity Laws)
old_vfd_savings_theoretical = 1 - (0.80 ** 3) # 48.8%
new_vfd_savings_theoretical = 1 - (0.60 ** 3) # 78.4%

# 2. REAL-WORLD ADJUSTMENT (Account for static head & efficiency drops)
# We apply a 20% penalty to the theoretical savings
real_world_factor = 0.80
old_vfd_savings_real = old_vfd_savings_theoretical * real_world_factor # ~39%
new_vfd_savings_real = new_vfd_savings_theoretical * real_world_factor # ~63%

print("--- REALISTIC VFD SAVINGS ---")
print(f"Old Plant Realistic Reduction: {old_vfd_savings_real * 100:.1f}%")
print(f"New Plant Realistic Reduction: {new_vfd_savings_real * 100:.1f}%")

# 3. NEW DAILY COST CALCULATION
baseline_daily_cost = 14817.00
old_plant_daily_cost = baseline_daily_cost * 0.60
new_plant_daily_cost = baseline_daily_cost * 0.40

new_old_cost = old_plant_daily_cost * (1 - old_vfd_savings_real)
new_new_cost = new_plant_daily_cost * (1 - new_vfd_savings_real)

realistic_daily_cost = new_old_cost + new_new_cost
realistic_daily_savings = baseline_daily_cost - realistic_daily_cost
realistic_annual_savings = realistic_daily_savings * 365

print(f"\nRealistic Daily Cost: ${realistic_daily_cost:,.2f}")
print(f"Realistic Annual Savings: ${realistic_annual_savings:,.2f}")

# 4. CONSERVATIVE TEA
capex_conservative = 16500000 # $16.5 Million
discount_rate = 0.08
years = 20

payback_years = capex_conservative / realistic_annual_savings
npv_realistic = -capex_conservative + sum((realistic_annual_savings / ((1 + discount_rate) ** year)) for year in range(1, years + 1))

print(f"\nConservative CAPEX: ${capex_conservative:,.2f}")
print(f"Realistic Payback Period: {payback_years:.1f} years")
print(f"Realistic Net Present Value (20 years): ${npv_realistic:,.2f}")

if npv_realistic > 0:
    print("CONCLUSION: The investment is FINANCIALLY VIABLE under conservative assumptions.")
else:
    print("CONCLUSION: The investment is NOT viable at this CAPEX.")
```
